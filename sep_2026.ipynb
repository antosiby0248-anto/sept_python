```python

# Diamond pattern


# Upper half
for i in range(4):
    for j in range(4 - i):
        print(" ", end=" ")
    for k in range(i + 1):
        print("*", end=" ")
    for m in range(i):
        print("*", end=" ")
    print()

# Lower half
for i in range(2, -1, -1):
    for j in range(4 - i):
        print(" ", end=" ")
    for k in range(i + 1):
        print("*", end=" ")
    for m in range(i):
        print("*", end=" ")
    print()

```

```python
class Student:
    def get_data(self):
        self.name = input("Enter the name: ")
        self.roll_no = int(input("Enter the roll no: "))

    def display_data(self):
        print(f"Name: {self.name}")
        print(f"Roll No: {self.roll_no}")


class Test(Student):
    def get_tmark(self):
        self.tmark = float(input("Enter test marks: "))

    def display_tmark(self):
        print(f"Test Marks: {self.tmark}")


class Sports(Student):
    def get_smark(self):
        self.smark = float(input("Enter sports marks: "))

    def display_smark(self):
        print(f"Sports Marks: {self.smark}")


class Grade(Test, Sports):
    def calgrade(self):
        # Calculate total marks by adding test and sports marks
        self.tomark = self.tmark + self.smark
        print(f"Total Marks: {self.tomark}")

        # Determine grade based on total marks
        if self.tomark > 80:
            print("Grade: A")
        elif self.tomark > 50:
            print("Grade: B")
        else:
            print("Status: Failed")


# --- Execution Example ---
g = Grade()

# Call methods to gather data
g.get_data()   # Inherited via Test/Sports from Student
g.get_tmark()  # Inherited from Test
g.get_smark()  # Inherited from Sports

print("\n--- Displaying Information ---")
# Call methods to display data
g.display_data()
g.display_tmark()
g.display_smark()
g.calgrade()
```

```python
class Student:
    def _init_(self, name, roll_no):
        self.name = name
        self.roll_no = roll_no

    def display_data(self):
        print(f"Name: {self.name}")
        print(f"Roll No: {self.roll_no}")


class Test(Student):
    def _init_(self, name, roll_no, tmark):
        # Explicitly initialize the Student parent class
        Student._init_(self, name, roll_no)
        self.tmark = tmark

    def display_tmark(self):
        print(f"Test Marks: {self.tmark}")


class Sports(Student):
    def _init_(self, name, roll_no, smark):
        # Explicitly initialize the Student parent class
        Student._init_(self, name, roll_no)
        self.smark = smark

    def display_smark(self):
        print(f"Sports Marks: {self.smark}")


# Grade inherits from both Test and Sports classes
class Grade(Test, Sports):
    def _init_(self, name, roll_no, tmark, smark):
        # Call base class constructors directly with positional variables
        Test._init_(self, name, roll_no, tmark)
        Sports._init_(self, name, roll_no, smark)

    def calgrade(self):
        self.tomark = self.tmark + self.smark
        print(f"Total Marks: {self.tomark}")

        if self.tomark > 80:
            print("Grade: A")
        elif self.tomark > 50:
            print("Grade: B")
        else:
            print("Status: Failed")


# --- Execution Example ---
g = Grade("athul", 2, 69.0, 80.0)

print("--- Displaying Information ---")
g.display_data()
g.display_tmark()
g.display_smark()
g.calgrade()
```
